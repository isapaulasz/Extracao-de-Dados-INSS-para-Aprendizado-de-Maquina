```python
import duckdb
from pathlib import Path

con = duckdb.connect()
for arq in sorted(Path("../../data/cat/raw").glob("*.parquet")):
    cols = [r[0] for r in con.execute(
        f"DESCRIBE SELECT * FROM read_parquet('{arq.as_posix()}')").fetchall()]
    print(arq.name, len(cols), cols, "\n")
```
