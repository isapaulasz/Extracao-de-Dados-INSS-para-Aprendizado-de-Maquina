# 02 · Qualidade dos dados (CAT)
Lê as tabelas geradas por `scripts/cat_05_auditoria.py` e produz as figuras do artigo. Rode a auditoria antes.

```python
import pandas as pd, matplotlib.pyplot as plt
from pathlib import Path
Q = Path('../../docs/qualidade'); FIG = Path('../../reports/cat/figures'); FIG.mkdir(parents=True, exist_ok=True)
rd = lambda n: pd.read_csv(Q/n, keep_default_na=False)
```

## % de vazios por coluna e por ano

```python
pv = pd.read_csv(Q/'cat_nulos_por_ano_pivot.csv').set_index('coluna')
pv = pv[pv.max(axis=1) > 0]
fig, ax = plt.subplots(figsize=(8, 0.28*len(pv)+1))
im = ax.imshow(pv.values, aspect='auto', cmap='viridis_r', vmin=0, vmax=100)
ax.set_xticks(range(pv.shape[1])); ax.set_xticklabels(pv.columns, rotation=45)
ax.set_yticks(range(len(pv))); ax.set_yticklabels(pv.index)
fig.colorbar(im, label='% vazio'); fig.tight_layout(); fig.savefig(FIG/'cat_vazios_por_ano.png', dpi=200)
```

## Distribuição por ano e tipo de CAT

```python
d = rd('cat_distribuicao_ano_tipo.csv').pivot(index='ano', columns='tipo_cat', values='n').fillna(0)
ax = d.plot(kind='bar', stacked=True, figsize=(8,4)); ax.set_ylabel('CATs'); plt.tight_layout(); plt.savefig(FIG/'cat_ano_tipo.png', dpi=200)
d
```

## CATs com benefício por arquivo (hipótese de datas de extração diferentes)

```python
a = rd('cat_distribuicao_arquivo.csv')
ax = a.plot(x='arquivo_origem', y='pct_com_beneficio', kind='bar', legend=False, figsize=(9,3.5)); ax.set_ylabel('% com benefício'); plt.tight_layout(); plt.savefig(FIG/'cat_pct_beneficio_arquivo.png', dpi=200)
a
```
